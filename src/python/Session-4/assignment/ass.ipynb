```python
def table():

    n = int(input("Enter number: "))

    for i in range(1, 11):
        print(n, "*", i, "=", n * i)

table()
```

```python
def prime(n):

    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


for i in range(3, 1000):

    if prime(i) and prime(i + 2):
        print(i, i + 2)
```

```python
def prime_factors():

    n = int(input("Enter number: "))

    i = 2

    while n > 1:

        if n % i == 0:
            print(i)
            n = n // i
        else:
            i = i + 1

prime_factors()
```

```python
def binary():

    n = int(input("Enter number: "))

    b = ""

    while n > 0:
        b = str(n % 2) + b
        n = n // 2

    print(b)

binary()
```

```python
def perfect():

    s = int(input("Enter start: "))
    e = int(input("Enter end: "))

    for n in range(s, e + 1):

        sum = 0

        for i in range(1, n):
            if n % i == 0:
                sum = sum + i

        if sum == n:
            print(n)

perfect()
```

```python
import os
import random

def thanos():

    folder = "Files"

    if not os.path.exists(folder):
        os.mkdir(folder)

    n = int(input("Enter number of files: "))

    for i in range(n):
        open(folder + "/file" + str(i) + ".txt", "w").close()

    files = os.listdir(folder)

    print("Files before delete =", len(files))

    random.shuffle(files)

    for i in range(len(files) // 2):
        os.remove(folder + "/" + files[i])

    files = os.listdir(folder)

    print("Files after delete =", len(files))

thanos()
```

```python
def add(x, y):
    return x + y

def sub(x, y):
    return x - y

def mul(x, y):
    return x * y

def div(x, y):
    if y == 0:
        return "Cannot divide by zero"
    return x / y


print("Welcome to the Simple Calculator!")

while True:

    print("\n1. Addition (+)")
    print("2. Subtraction (-)")
    print("3. Multiplication (*)")
    print("4. Division (/)")
    print("Type 'exit' to quit")

    ch = input("Enter your choice: ")

    if ch == "exit":
        print("Thank you! Goodbye.")
        break

    if ch not in ["1", "2", "3", "4"]:
        print("Invalid choice")
        continue

    try:
        n1 = float(input("Enter first number: "))
        n2 = float(input("Enter second number: "))
    except:
        print("Invalid input")
        continue

    if ch == "1":
        print("Result =", add(n1, n2))

    elif ch == "2":
        print("Result =", sub(n1, n2))

    elif ch == "3":
        print("Result =", mul(n1, n2))

    elif ch == "4":
        print("Result =", div(n1, n2))
```
